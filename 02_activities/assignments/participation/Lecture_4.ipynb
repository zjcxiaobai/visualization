```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import pandas
import PIL
import requests
```

```python
np.random.seed(0)  # For reproducibility
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)
x2 = np.array(["Luffy ", "Zoro ", "Nami ", "Usopp ", "Sanji "])
y2 = np.array([110,180,240, 99,220])
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7, 3))
ax1.scatter(x1, y1)
ax1.annotate('This is important', xy = (10, 65), xytext=(30, 65), arrowprops = dict(facecolor='black'))
ax2.bar(x2, y2)
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7,4), layout='constrained')
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2, y2)
someaxes['ax3'].plot(x1,y1)
someaxes['ax1'].set_xlabel('A BIG LABEL', fontsize = 18)
someaxes['ax2'].set_xlabel('Another BIG LABEL', fontsize = 18)
someaxes['ax3'].set_xlabel('label 2', fontsize = 18)
```

```python
x = np.array(["Luffy ", "Zoro ", "Nami ", "Usopp ", "Sanji "])
y1 = np.array([110,180,240, 99,220])
y2 = np.array([170,100,90, 99,50])

fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(x,y1, color = 'indigo')  
ax.plot(x,y2, color = 'orange')
```

```python
y2_std = np.std(y2)
fig,ax = plt.subplots(figsize=(7,3))
# ax.plot(x,y2, color = 'orange')
ax.errorbar(x,y2, yerr=y2_std, color = 'orange', elinewidth=3, capsize=10, capthick=4, errorevery=3)
```
