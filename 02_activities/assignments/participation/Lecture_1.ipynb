```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import pandas
import PIL
import requests
```

# Making a basic figure with matplotlib

```python
np.random.seed(42)  # For reproducibility
X = np.arange(50)
Y = np.random.randint(0, 100, size=50)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(X,
           Y,
           marker = '*',
           color = 'indigo')
plt.show()
```

```python
# bar plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(X, Y)
```

```python
# line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(X, Y, 
        marker = '*',
        markersize = 12,
        # color = 'indigo',
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')
ax.grid(axis = 'y', color = 'blue', linestyle = '-.', linewidth = 0.5)
```

```python
# histogram
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(Y, bins=10)
ax.set_title("Total growth over time", fontdict=font1)
ax.set_ylabel("Total growth", fontdict=font2)
ax.set_xlabel("Years since start", fontdict=font2)
fig.tight_layout()
```

# Python Graph Activity

```python
# libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
 
# create data
size = 100000
df = pd.DataFrame({
   'x': np.random.normal(size=size),
   'y': np.random.normal(size=size)
})
df.head()
```

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(8,8))
x = df['x']
y = df['y']
# Big bins
axs[0,0].hist2d(x, y, bins=(50, 50), cmap=plt.cm.jet)
axs[0, 0].set_title('bins = (50, 50)')
 
# Small bins
axs[0,1].hist2d(x, y, bins=(600, 600), cmap=plt.cm.jet)
axs[0, 1].set_title('bins = (600, 600)')
 
# If you do not set the same values for X and Y, the bins won't be a square!
axs[1,0].hist2d(x, y, bins=(600, 30), cmap=plt.cm.jet)
axs[1, 0].set_title('bins = (600, 30)')

# If you do not set the same values for X and Y, the bins won't be a square!
axs[1,1].hist2d(x, y, bins=(30, 600), cmap=plt.cm.jet)
axs[1, 1].set_title('bins = (30, 600)')

plt.show()
```

This is good way to visualize where do data points pile up, with more bins, the results smooth out, less bins we see more obvious boundries.
