```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
import pandas
import PIL
import requests
import seaborn as sns
```

```python
from PIL import Image
import requests
from io import BytesIO
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {'User-Agent': 'Mozilla/5.0', "Accept":"image/*, */*;q=0.8", "referer":"https://en.wikipedia.org/"}
response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
img = Image.open(image_file)
img
```

```python
x = np.array(["Luffy ", "Zoro ", "Nami ", "Usopp ", "Sanji "])
y1 = np.array([110,180,240, 99,220])
y2 = np.array([170,100,90, 120,50])

fig, ax= plt.subplots(figsize = (7,3))
ax.plot(x, y2, color = 'red')
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])
ax_image.imshow(img)
ax_image.axis('off')
path = '.'
filename = '/fig1a.png'
plt.savefig(path + filename, dpi=300)
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.set_style("ticks")
tipgraph = sns.lineplot(data=tips, x="total_bill", y="tip",
                        color = "purple",
                        linestyle = "--",
                        marker = "o",
                        markerfacecolor = "orange",)
tipgraph.set(title = "Tips vs. Total Bill",
                   xlabel = "Total Bill ($)",
                   ylabel = "Tip Amount ($)")
```

```python
tipgraph = sns.scatterplot(data=tips, x="total_bill", y="tip",
                           style='time',
                            hue='day',
                            palette = ['purple', 'orange', 'green', 'blue']
)
tipgraph.set(title = "Tips vs. Total Bill",
                   xlabel = "Total Bill ($)",
                   ylabel = "Tip Amount ($)")
```

```python
sns.pairplot(data = tips, hue = 'day')
```

```python
daysplot = sns.relplot(data=tips, x="total_bill", y="tip", col="day", hue = 'sex', kind="scatter", col_wrap = 2)
```

```python
import plotly.graph_objects as go
x1 = np.array(["Luffy ", "Zoro ", "Nami ", "Usopp ", "Sanji "])
y1 = np.array([110,180,240, 99,220])
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(title="Pirate Scores",
                     xaxis_title="Pirates",
                     yaxis_title="Scores")
graph.show()
```

```python
graph.write_html("pirate_scores.html")
```

```python
graph_1 = go.Figure()
graph_1.add_scatter(x=x1, y=y1, mode='markers',
                  marker = dict(size=12, color='hotpink', opacity=1,
                                line=dict(width=2, color='purple')) 
)
graph_1.update_layout(title="Interactive Pirate plot",
                    xaxis_title="Pirates",
                    yaxis_title="Scores",
                    width=500,
                    height=500)
```

```python
from wordcloud import WordCloud
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv', on_bad_lines='skip')
print(df)
```

```python
text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color='white', colormap='viridis').generate(text)
fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(['apple', 'banana', 'cherry', 'date'])
B = set(['cherry', 'date', 'fig', 'grape'])
diagram = venn2([A, B], set_labels = ('Friut', 'Pie'), alpha = 0.7,
                set_colors = ('purple', 'orange'))
```

```python
diagram.get_label_by_id('10').set_text('n'.join(A- B))  
diagram.get_label_by_id('11').set_text('n'.join(B & A))
diagram.get_label_by_id('01').set_text('n'.join(B - A))
```
